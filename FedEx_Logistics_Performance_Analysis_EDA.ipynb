## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# Convert date columns
date_cols = [
    'PQ First Sent to Client Date',
    'PO Sent to Vendor Date',
    'Scheduled Delivery Date',
    'Delivered to Client Date',
    'Delivery Recorded Date'
]

for col in date_cols:
    df[col] = pd.to_datetime(df[col], errors='coerce', dayfirst=True)

# Convert mixed numeric columns
numeric_cols = [
    'Line Item Quantity',
    'Line Item Value',
    'Pack Price',
    'Unit Price',
    'Weight (Kilograms)',
    'Freight Cost (USD)',
    'Line Item Insurance (USD)'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Delivery performance
df['Delay_Days'] = (
    df['Delivered to Client Date'] - df['Scheduled Delivery Date']
).dt.days

df['Delivery_Status'] = np.where(
    df['Delay_Days'] <= 0, 'On Time', 'Delayed'
)

# Planned lead time
df['Lead_Time_Days'] = (
    df['Scheduled Delivery Date'] - df['PO Sent to Vendor Date']
).dt.days

print("Cleaning completed.")
````

### What all manipulations have you done and insights you found?

Some columns contain numbers mixed with text such as operational notes. For analysis, only values that can be converted to numeric are used. Date columns are converted with invalid entries becoming missing dates.

We also create two useful business measures:

- **Delay_Days** = actual delivery date − scheduled delivery date
- **Delivery_Status** = On Time when actual delivery is on or before the scheduled date
